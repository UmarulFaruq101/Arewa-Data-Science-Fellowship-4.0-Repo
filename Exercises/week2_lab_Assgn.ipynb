**Q1.** What is a literal, and how is it different from a variable?

answer:

A literal is a fixed value typed directly into the code, such as `42` or `"hello"`. A variable is a name that points to a value stored in memory, and that value can change while the program runs.

**Q2.** `int("42")` works but `int("3.7")` raises an error. Explain why.

answer:

`int()` only knows how to read strings that represent whole numbers. `"42"` is a valid whole number, but `"3.7"` contains a decimal point, which `int()` does not know how to parse, so it raises a `ValueError`.

**Q3.** Give the value and the type of `7 / 2`, `7 // 2`, and `7 % 2`.

answer:

`7 / 2` is `3.5`, a `float`. `7 // 2` is `3`, an `int`. `7 % 2` is `1`, an `int` (the remainder after dividing 7 by 2).

**Q4.** Why is `print("Total: " + str(total))` necessary, when `print("Total:", total)` works without `str`?

answer:

The `+` operator joins two strings together, so both sides must already be strings, which is why `total` needs `str()` first. The comma version passes `"Total:"` and `total` as two separate arguments to `print`, and `print` converts each argument to text and adds a space between them automatically.

**Q5.** `0.1 + 0.2 == 0.3` is `False`. What does this tell you about the `float` type, and what should you do instead when comparing floats?
answer:
It shows that floats are stored as approximations in binary, so tiny rounding errors can creep in and make two mathematically equal values compare as unequal. Instead of `==`, you should check whether the difference between the two values is smaller than a small tolerance, e.g. `abs(a - b) < 1e-9`.

**Q6.** Give three valid variable names and three invalid ones, and state which rule each invalid one breaks.

**Q7.** A bug found during maintenance costs far more to fix than one found during analysis. Using the software development process from the slides, explain in your own words why.

answer:

Once a program is in maintenance, the original design decisions, code, and documentation may have been written months or years earlier, other features and code now depend on the buggy part, and the fix has to be re-tested against everything that was built on top of it. Catching the same bug during analysis means changing a plan on paper, before any of that dependent work exists.

**Q8.** What is the difference between a docstring and an end-of-line comment, and when would you use each?

answer:

A docstring is a triple-quoted block at the top of a program or function that explains its overall purpose, inputs, and outputs. An end-of-line comment is a short note after `#` on a single line, used to clarify one specific piece of code, such as what a constant represents.
