## Question 1: What is the difference between `append` and `extend`?

**Answer:** `append` adds **one item** to the end of a list, and that item is added as a single element, even if it is itself a list. `extend` takes an **iterable** (such as a list) and adds **each of its items** to the end of the list, one by one.

The demonstration below shows the difference.

```python
# append adds ONE element (here, the whole list [3, 4] becomes a single item)
with_append = [1, 2]
with_append.append([3, 4])
print("After append:", with_append)
print("Length:", len(with_append))

# extend adds EACH element of the iterable separately
with_extend = [1, 2]
with_extend.extend([3, 4])
print("After extend:", with_extend)
print("Length:", len(with_extend))
```

## Question 2: How can you determine an element's presence or absence in a list?

**Answer:** Use the **`in` operator**. The expression `x in my_list` gives `True` if `x` is in the list and `False` if it is not. Use `not in` to test for absence.

```python
scores = [34, 45, 67]

print(45 in scores)       # True  (present)
print(99 in scores)       # False (absent)
print(99 not in scores)   # True  (absence test)
```

## Question 3: What is the difference between object identity and structural equivalence?

**Answer:**
- **Object identity** asks: *are these two names pointing to the very same object in memory?* It is tested with `is`.
- **Structural equivalence** asks: *do these two objects have equal contents?* It is tested with `==`.

Two different lists can be structurally equivalent (`==` is `True`) without being the same object (`is` is `False`).

```python
original = [20, 30, 40]
alias = original            # same object, a second name for it
copy = list(original)       # a new, separate object with the same contents

print("original == alias :", original == alias)   # True  (same contents)
print("original is alias :", original is alias)   # True  (same object)
print("original == copy  :", original == copy)    # True  (same contents)
print("original is copy  :", original is copy)    # False (different objects)
```

## Task B: Structures in other languages that are similar to Python dictionaries

| Language | Similar structure |
|---|---|
| JavaScript | Objects (`{key: value}`) and `Map` |
| Java | `HashMap` / `Map` |
| C++ | `std::map` and `std::unordered_map` |
| PHP | Associative arrays |
| R | Named lists |
| C# | `Dictionary<TKey, TValue>` |
| Ruby | `Hash` |

**How they differ from Python dictionaries:**
- In **Java, C++ and C#**, the key and value types are usually declared in advance (for example, all keys must be strings). A Python dictionary can mix key types and value types freely, as long as each key is immutable.
- In **JavaScript objects**, keys are normally strings, while Python keys can be strings, numbers, or tuples.
- Python dictionaries **keep insertion order** (from Python 3.7). Some other languages, such as a C++ `unordered_map`, do not guarantee any order.
- The syntax differs. Python uses `d[key]` and `d.get(key, default)`, while other languages use their own methods.

## 2. Similarities and differences among lists, tuples, and dictionaries

| Feature | List | Tuple | Dictionary |
|---|---|---|---|
| Brackets | `[ ]` | `( )` | `{ }` |
| Example | `[1, 2, 3]` | `(1, 2, 3)` | `{"a": 1, "b": 2}` |
| Ordered | Yes | Yes | Yes (by insertion order) |
| Changeable (mutable) | Yes | **No** (immutable) | Yes |
| Accessed by | Position (index) | Position (index) | Key |
| Duplicates allowed | Yes | Yes | Keys must be **unique**; values can repeat |
| Typical use | Collections that change | Fixed records | Lookups and associations |

**Similarities:**
- All three can hold items of different types.
- All three work with `len()` and can be looped over with `for`.
- Lists and tuples both support indexing and slicing.

**Differences:**
- Tuples **cannot** be modified once created, while lists and dictionaries can.
- Lists and tuples use **positions**, while dictionaries use **keys**.

The code below demonstrates the main difference, which is mutability.

```python
my_list = [1, 2, 3]
my_list[0] = 100          # works: lists are mutable
print("List after change:", my_list)

my_tuple = (1, 2, 3)
try:
    my_tuple[0] = 100     # fails: tuples are immutable
except TypeError as error:
    print("Tuple error:", error)

my_dict = {"a": 1, "b": 2}
my_dict["a"] = 100        # works: dictionaries are mutable
print("Dictionary after change:", my_dict)
```

## Question 1: Create a list of five integers and replace the middle element with `0`

## Question 2: Start with `[3, 1, 4]` and use only `append`, `extend`, and `sort` to obtain `[1, 3, 4, 5, 9]`

**Plan:**
1. `append(5)` adds the single item 5 to the end.
2. `extend([9])` adds the items of `[9]` to the end.
3. `sort()` puts everything in order.

The list is now `[3, 1, 4, 5, 9]`, and sorting gives `[1, 3, 4, 5, 9]`.

```python
data = [3, 1, 4]
print("Start:        ", data)

data.append(5)         # [3, 1, 4, 5]
print("After append: ", data)

data.extend([9])       # [3, 1, 4, 5, 9]
print("After extend: ", data)

data.sort()            # [1, 3, 4, 5, 9]
print("After sort:   ", data)

print("Matches target:", data == [1, 3, 4, 5, 9])
```

## Question 3: Write a function `is_member(x, xs)` without using `in`

The function loops over the list and compares each item with `x`. If a match is found it returns `True` immediately. If the loop finishes with no match, it returns `False`.

```python
def is_member(x, xs):
    """Return True if x is in the list xs, otherwise False (without using 'in' as a test)."""
    for item in xs:
        if item == x:
            return True
    return False


# Tests
print(is_member(3, [1, 2, 3, 4]))        # True
print(is_member(9, [1, 2, 3, 4]))        # False
print(is_member("a", ["a", "b"]))        # True
print(is_member(1, []))                  # False (empty list)
```

## Question 4: Build a dictionary `counts` that maps each character in `"banana"` to its frequency

**Method:** loop over each character. If it is already a key, add 1 to its count. If not, start it at 1. `counts.get(char, 0)` handles both cases, because it returns `0` when the key does not exist yet.

```python
word = "banana"
counts = {}

for char in word:
    counts[char] = counts.get(char, 0) + 1

print(counts)
```

## Question 5: What happens with `d2 = d` and then `d["a"] = 99`? How do you avoid aliasing?

**Answer:** `d2 = d` does **not** make a copy. It makes `d2` a second name for the **same** dictionary object (aliasing). So when `d["a"]` is changed to `99`, `d2["a"]` also shows `99`, because both names point to one dictionary.

**How to avoid it:** make a real copy with `d.copy()` or `dict(d)`. The copy is a separate object, so later changes to `d` do not affect it.

```python
d = {"a": 1, "b": 2}

# --- Aliasing (NOT a copy) ---
d2 = d
d["a"] = 99
print("d  :", d)
print("d2 :", d2)                 # d2 also changed
print("Same object?", d is d2)    # True

print()

# --- Avoiding aliasing with a real copy ---
d = {"a": 1, "b": 2}
d3 = d.copy()                     # or dict(d)
d["a"] = 99
print("d  :", d)
print("d3 :", d3)                 # d3 is unchanged
print("Same object?", d is d3)    # False
```
