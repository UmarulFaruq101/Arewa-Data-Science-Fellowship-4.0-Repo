**18.2 TODO, INTERACTIVE.** The guessing game from the slides. The computer picks a number, you guess, and it answers `Too small`, `Too large`, or congratulates you and stops. Complete the body of the loop.

```python
import random

smaller = int(input("Enter the smaller number: "))
larger = int(input("Enter the larger number: "))
myNumber = random.randint(smaller, larger)
count = 0

while True:
    count += 1
    userNumber = int(input("Enter your guess: "))
    if userNumber < myNumber:
        print("Too small")
    elif userNumber > myNumber:
        print("Too large")
    else:
        print("Correct! You got it in", count, "guesses.")
        break
```

### Exercise 3. FizzBuzz

```python
output = []

for number in range(1, 31):
    if number % 3 == 0 and number % 5 == 0:
        output.append("FizzBuzz")
    elif number % 3 == 0:
        output.append("Fizz")
    elif number % 5 == 0:
        output.append("Buzz")
    else:
        output.append(number)

print(output)
```

```python
# CHECK
_expected = ["FizzBuzz" if k % 15 == 0 else "Fizz" if k % 3 == 0 else "Buzz" if k % 5 == 0 else k for k in range(1, 31)]
assert len(output) == 30, "output should have 30 entries, you have " + str(len(output))
for _k in range(30):
    assert output[_k] == _expected[_k], "entry for the number " + str(_k + 1) + " should be " + repr(_expected[_k]) + ", you got " + repr(output[_k])
print("Looks good.")
```

### Exercise 4. Digit sum

### Exercise 6. Grade statistics

```python
scores = [72, 45, 88, 50, 91, 33, 67, 49, 100, 58]

highest = scores[0]
lowest = scores[0]
total = 0
passCount = 0

for score in scores:
    if score > highest:
        highest = score
    if score < lowest:
        lowest = score
    total += score
    if score >= 50:
        passCount += 1

average = total / len(scores)
print(highest, lowest, average, passCount)
```

```python
# CHECK
assert highest == 100, "highest should be 100, you got " + repr(highest)
assert lowest == 33, "lowest should be 33, you got " + repr(lowest)
assert abs(average - 65.3) < 1e-9, "average should be 65.3, you got " + repr(average)
assert passCount == 7, "7 scores are 50 or above (remember that 50 itself is a pass), you got " + repr(passCount)
print("Looks good.")
```

### Exercise 7. Challenge, no automatic check

```python
"""
Program: Number Guessing Game
Author: Umar
Date: 2026-09-25
Purpose: let the player guess a computer-chosen number between 1 and 100.
Input: the player's guesses, typed in one at a time.
Output: "Too small" / "Too large" after each guess, or the answer if guesses run out.
"""

import random

LOWER_BOUND = 1
UPPER_BOUND = 100
MAX_GUESSES = 7

secretNumber = random.randint(LOWER_BOUND, UPPER_BOUND)
guessesLeft = MAX_GUESSES

print(f"Guess a number between {LOWER_BOUND} and {UPPER_BOUND}. You have {MAX_GUESSES} guesses.")

while guessesLeft > 0:
    guess = int(input("Enter your guess: "))

    # reject out-of-range input without spending a guess
    if guess < LOWER_BOUND or guess > UPPER_BOUND:
        print(f"Please guess between {LOWER_BOUND} and {UPPER_BOUND}.")
        continue

    guessesLeft -= 1

    if guess < secretNumber:
        print("Too small")
    elif guess > secretNumber:
        print("Too large")
    else:
        print(f"Correct! You got it with {guessesLeft} guesses to spare.")
        break
else:
    print(f"Out of guesses. The number was {secretNumber}.")
```
