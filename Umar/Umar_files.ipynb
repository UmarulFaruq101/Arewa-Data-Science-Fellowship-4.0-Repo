```python
import pandas as pd

import numpy as np

import matplotlib.pyplot as plt

import seaborn as sns

from rdkit import Chem

from rdkit.Chem import AllChem, Descriptors, rdMolDescriptors

from rdkit.Chem.MolStandardize import rdMolStandardize

from sklearn.model_selection import train_test_split

from sklearn.preprocessing import StandardScaler

from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error

from sklearn.feature_selection import VarianceThreshold

from sklearn.impute import SimpleImputer

from sklearn.linear_model import (

    LinearRegression, Ridge, Lasso, ElasticNet, BayesianRidge,

    SGDRegressor, HuberRegressor, RANSACRegressor, TweedieRegressor

)

from sklearn.svm import SVR, LinearSVR

from sklearn.tree import DecisionTreeRegressor

from sklearn.ensemble import (

    RandomForestRegressor, GradientBoostingRegressor, AdaBoostRegressor,

    BaggingRegressor, ExtraTreesRegressor

)

from sklearn.neighbors import KNeighborsRegressor

from sklearn.neural_network import MLPRegressor

import joblib

import os

import warnings

warnings.filterwarnings('ignore')





class Chemical_Property_Predictor:

    def __init__(self, output_dir='best_models'):

        self.raw_data = None

        self.data = None

        self.temp_data = None

        self.descriptors = None

        self.X = None

        self.y = None

        self.X_train = None

        self.X_test = None

        self.y_train = None

        self.y_test = None

        self.scaler = None

        self.imputer = None

        self.model = None

        self.best_model = None

        self.best_model_name = None

        self.best_r2 = -np.inf

        self.output_dir = output_dir

        self.feature_names = None

        self.X_ = None

        

        os.makedirs(self.output_dir, exist_ok=True)

        

        # Stable models (no problematic ones)

        self.models = {

            'Linear Regression': LinearRegression(),

            'Ridge Regression': Ridge(),

            'Lasso Regression': Lasso(),

            'ElasticNet': ElasticNet(),

            'Bayesian Ridge': BayesianRidge(),

            'SGD Regressor': SGDRegressor(max_iter=1000, tol=1e-3, random_state=42),

            'Huber Regressor': HuberRegressor(),

            'RANSAC': RANSACRegressor(random_state=42),

            'Tweedie': TweedieRegressor(power=0, max_iter=1000),

            'SVR': SVR(),

            'Linear SVR': LinearSVR(max_iter=10000, random_state=42),

            'Decision Tree': DecisionTreeRegressor(random_state=42),

            'Random Forest': RandomForestRegressor(n_estimators=100, random_state=42),

            'Gradient Boosting': GradientBoostingRegressor(n_estimators=100, random_state=42),

            'AdaBoost': AdaBoostRegressor(n_estimators=100, random_state=42),

            'Bagging': BaggingRegressor(n_estimators=100, random_state=42),

            'Extra Trees': ExtraTreesRegressor(n_estimators=100, random_state=42),

            'K-Neighbors': KNeighborsRegressor(),

            'Neural Network': MLPRegressor(hidden_layer_sizes=(10, 5), max_iter=1000, random_state=42)

        }



    def _process_molecule(self, smiles):

        try:

            mol = Chem.MolFromSmiles(smiles)

            if mol is None:

                return None

            mol = Chem.MolFromSmiles(Chem.MolToSmiles(mol))

            if mol is None:

                return None

            parent = rdMolStandardize.FragmentParent(mol)

            mol = Chem.AddHs(parent)

            try:

                AllChem.EmbedMolecule(mol, randomSeed=42)

                AllChem.MMFFOptimizeMolecule(mol)

            except:

                pass

            return mol

        except:

            return None



    def load_data(self, file_path):

        """Load CSV or Excel file and auto‑rename columns to expected names."""

        if file_path.endswith(('.xlsx', '.xls')):

            self.raw_data = pd.read_excel(file_path)

        else:

            self.raw_data = pd.read_csv(file_path)

        

        # Map actual column names to expected ones

        rename_map = {}

        if 'NAME' in self.raw_data.columns:

            rename_map['NAME'] = 'substance_name'

        if 'SMILES' in self.raw_data.columns:

            rename_map['SMILES'] = 'smiles'

        if 'Log_Psat_mmHg' in self.raw_data.columns:

            rename_map['Log_Psat_mmHg'] = 'property'

        # Temperature columns already have correct names, no rename needed

        

        if rename_map:

            self.raw_data = self.raw_data.rename(columns=rename_map)

        

        # Now check all required columns exist

        required = ['substance_name', 'smiles', 'property', 'Inverse_Temperature_K', 'Temperature_C']

        for col in required:

            if col not in self.raw_data.columns:

                raise ValueError(f"Column '{col}' not found. Found: {list(self.raw_data.columns)}")

        

        self.data = self.raw_data[['substance_name', 'smiles', 'property']].copy()

        self.temp_data = self.raw_data[['Inverse_Temperature_K', 'Temperature_C']].copy()

        print(f"Loaded {len(self.data)} rows from {file_path}")



    def generate_descriptors(self):

        if self.data is None:

            raise ValueError("Load data first.")

        descriptor_list = []

        temp_list = []

        valid_indices = []

        for idx, row in self.data.iterrows():

            mol = self._process_molecule(row['smiles'])

            if mol is not None:

                desc = {}

                for name, func in Descriptors.descList:

                    try:

                        desc[name] = func(mol)

                    except:

                        desc[name] = np.nan

                desc['NumAtoms'] = mol.GetNumAtoms()

                desc['NumHeavyAtoms'] = mol.GetNumHeavyAtoms()

                desc['NumRotatableBonds'] = rdMolDescriptors.CalcNumRotatableBonds(mol)

                desc['NumHDonors'] = rdMolDescriptors.CalcNumHBD(mol)

                desc['NumHAcceptors'] = rdMolDescriptors.CalcNumHBA(mol)

                desc['TPSA'] = rdMolDescriptors.CalcTPSA(mol)

                desc['MolLogP'] = Descriptors.MolLogP(mol)

                desc['MolWt'] = Descriptors.MolWt(mol)

                desc['RingCount'] = rdMolDescriptors.CalcNumRings(mol)

                try:

                    desc['PMI1'] = rdMolDescriptors.CalcPMI1(mol)

                    desc['PMI2'] = rdMolDescriptors.CalcPMI2(mol)

                    desc['PMI3'] = rdMolDescriptors.CalcPMI3(mol)

                    desc['RadiusOfGyration'] = rdMolDescriptors.CalcRadiusOfGyration(mol)

                except:

                    pass

                descriptor_list.append(desc)

                temp_list.append(self.temp_data.loc[idx].values)

                valid_indices.append(idx)

            else:

                print(f"Could not process SMILES: {row['smiles']}")

        self.descriptors = pd.DataFrame(descriptor_list, index=valid_indices)

        self.temp_data = pd.DataFrame(temp_list, 

                                      columns=['Inverse_Temperature_K', 'Temperature_C'],

                                      index=valid_indices)

        self.data = self.data.loc[valid_indices]

        print(f"Generated {self.descriptors.shape[1]} molecular descriptors + 2 temperature features.")



    def preprocess_data(self, test_size=0.2, random_state=42, strategy='mean'):

        if self.descriptors is None:

            raise ValueError("Generate descriptors first.")

        X_combined = pd.concat([self.descriptors, self.temp_data], axis=1)

        selector = VarianceThreshold(threshold=0.001)

        selected = selector.fit_transform(X_combined)

        self.feature_names = np.array(X_combined.columns)[selector.get_support()]

        X_filtered = pd.DataFrame(selected, columns=self.feature_names)

        for temp_col in ['Inverse_Temperature_K', 'Temperature_C']:

            if temp_col not in self.feature_names:

                X_filtered[temp_col] = X_combined[temp_col].values

                self.feature_names = list(self.feature_names) + [temp_col]

        self.X = X_filtered

        self.y = self.data['property']

        self.X_train, self.X_test, self.y_train, self.y_test = train_test_split(

            self.X, self.y, test_size=test_size, random_state=random_state)

        self.imputer = SimpleImputer(strategy=strategy)

        self.scaler = StandardScaler()

        self.X_train = pd.DataFrame(

            self.scaler.fit_transform(self.imputer.fit_transform(self.X_train)),

            columns=self.feature_names

        )

        self.X_test = pd.DataFrame(

            self.scaler.transform(self.imputer.transform(self.X_test)),

            columns=self.feature_names

        )

        self.X_ = self.X

        print(f"Train: {len(self.y_train)}, Test: {len(self.y_test)}")



    def evaluate_all_models(self):

        print("\n=== Evaluating All Models (with Temperature) ===")

        results = []

        for name, model in self.models.items():

            try:

                print(f"\nTraining {name}...")

                model.fit(self.X_train, self.y_train)

                pred = model.predict(self.X_test)

                r2 = r2_score(self.y_test, pred)

                rmse = np.sqrt(mean_squared_error(self.y_test, pred))

                results.append((name, r2, rmse))

                if r2 > self.best_r2:

                    self.best_r2 = r2

                    self.best_model = model

                    self.best_model_name = name

                    print(f"  >>> New best: {name} (R²={r2:.3f})")

            except Exception as e:

                print(f"  {name} failed: {str(e)[:80]}...")

        results.sort(key=lambda x: x[1], reverse=True)

        print("\n=== Final Rankings ===")

        print("{:<25} {:<10} {:<10}".format("Model", "R²", "RMSE"))

        for name, r2, rmse in results:

            print("{:<25} {:<10.3f} {:<10.3f}".format(name, r2, rmse))

        self.save_best_model()

        return results



    def save_best_model(self):

        if self.best_model is None:

            print("No best model found.")

            return

        filename = f"best_model_{self.best_model_name.replace(' ', '_')}.pkl"

        filepath = os.path.join(self.output_dir, filename)

        joblib.dump({

            'model': self.best_model,

            'scaler': self.scaler,

            'imputer': self.imputer,

            'feature_names': self.feature_names,

            'model_name': self.best_model_name,

            'test_r2': self.best_r2

        }, filepath)

        print(f"\nBest model saved to: {filepath}")



    def plot_parity(self):

        if self.best_model is None:

            print("Train a model first.")

            return

        train_pred = self.best_model.predict(self.X_train)

        test_pred = self.best_model.predict(self.X_test)

        plt.figure(figsize=(8,6))

        plt.scatter(self.y_train, train_pred, alpha=0.5, label='Train')

        plt.scatter(self.y_test, test_pred, alpha=0.5, label='Test')

        lims = [min(self.y.min(), train_pred.min(), test_pred.min()),

                max(self.y.max(), train_pred.max(), test_pred.max())]

        plt.plot(lims, lims, 'k--')

        plt.xlabel('True Property')

        plt.ylabel('Predicted Property')

        plt.title(f'Parity Plot – {self.best_model_name} (with Temperature)')

        plt.legend()

        plt.grid()

        plt.show()



    def plot_feature_importance(self, top_n=20):

        if self.best_model is None:

            print("No model available.")

            return

        if hasattr(self.best_model, 'coef_'):

            imp = np.abs(self.best_model.coef_)

        elif hasattr(self.best_model, 'feature_importances_'):

            imp = self.best_model.feature_importances_

        else:

            print("No feature importance available.")

            return

        idx = np.argsort(imp)[::-1][:top_n]

        plt.figure(figsize=(10,6))

        plt.barh(range(len(idx)), imp[idx][::-1])

        plt.yticks(range(len(idx)), [self.feature_names[i] for i in idx][::-1])

        plt.xlabel('Importance')

        plt.title(f'Top {top_n} Features – {self.best_model_name}')

        plt.tight_layout()

        plt.show()


if __name__ == "__main__":

    predictor = Chemical_Property_Predictor()

    predictor.load_data('Original_data.xlsx')   # <-- use your file name

    predictor.generate_descriptors()

    predictor.preprocess_data()

    predictor.evaluate_all_models()

    predictor.plot_parity()

    predictor.plot_feature_importance() 

```

```python
# -*- coding: utf-8 -*-
"""
Created on Sun Jun 28 15:03:58 2026

@author: user
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from rdkit import Chem
from rdkit.Chem import AllChem, Descriptors, rdMolDescriptors
from rdkit.Chem.MolStandardize import rdMolStandardize
from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV, RandomizedSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
from sklearn.feature_selection import VarianceThreshold
from sklearn.impute import SimpleImputer
from sklearn.pipeline import make_pipeline
from sklearn.neural_network import MLPRegressor
import joblib
import os
import warnings
warnings.filterwarnings('ignore')

class Chemical_Property_Predictor_NN:
    def __init__(self, output_dir='best_models'):
        self.raw_data = None
        self.data = None
        self.temp_data = None
        self.descriptors = None
        self.X = None
        self.y = None
        self.X_train = None
        self.X_test = None
        self.y_train = None
        self.y_test = None
        self.scaler = None
        self.imputer = None
        self.model = None
        self.best_model = None
        self.best_model_name = None
        self.best_r2 = -np.inf
        self.best_params = None
        self.output_dir = output_dir
        self.feature_names = None
        
        # Create output directory if it doesn't exist
        os.makedirs(self.output_dir, exist_ok=True)
        
        # Initialize standardizer for salt stripping
        self.normalizer = rdMolStandardize.Normalizer()
        self.uncharger = rdMolStandardize.Uncharger()
        
    def _process_molecule(self, smiles):
        """Process a single molecule through all steps"""
        try:
            # 1. Read SMILES
            mol = Chem.MolFromSmiles(smiles)
            if mol is None:
                return None
            
            # 2. Transform to molecule and sanitize
            mol = Chem.MolFromSmiles(Chem.MolToSmiles(mol))
            if mol is None:
                return None
            
            # 3. Strip salts and standardize
            parent_mol = rdMolStandardize.FragmentParent(mol)
            
            # 4. Add hydrogens
            mol = Chem.AddHs(parent_mol)
            
            # 5. Generate coordinates
            try:
                AllChem.EmbedMolecule(mol, randomSeed=42)
                AllChem.MMFFOptimizeMolecule(mol)
            except:
                pass
            
            return mol
            
        except Exception as e:
            print(f"Error processing molecule {smiles}: {str(e)}")
            return None
    
    def load_data(self, file_path):
        """Load chemical data from CSV or Excel file, handling name mappings and Inverse_Temperature_K"""
        if file_path.endswith(('.xlsx', '.xls')):
            self.raw_data = pd.read_excel(file_path)
        else:
            self.raw_data = pd.read_csv(file_path)
        
        rename_map = {}
        if 'NAME' in self.raw_data.columns:
            rename_map['NAME'] = 'substance_name'
        if 'SMILES' in self.raw_data.columns:
            rename_map['SMILES'] = 'smiles'
        if 'Log_Psat_mmHg' in self.raw_data.columns:
            rename_map['Log_Psat_mmHg'] = 'property'
            
        if rename_map:
            self.raw_data = self.raw_data.rename(columns=rename_map)
            
        required = ['substance_name', 'smiles', 'property', 'Inverse_Temperature_K']
        for col in required:
            if col not in self.raw_data.columns:
                raise ValueError(f"Column '{col}' not found. Available columns: {list(self.raw_data.columns)}")
        
        self.data = self.raw_data[['substance_name', 'smiles', 'property']].copy()
        self.temp_data = self.raw_data[['Inverse_Temperature_K']].copy()
        print(f"Loaded {len(self.data)} rows from {file_path}")

    def generate_descriptors(self):
        """Generate molecular descriptors using RDKit with full processing and retain Inverse_Temperature_K"""
        if self.data is None:
            raise ValueError("No data loaded. Please load data first.")
            
        descriptor_list = []
        temp_list = []
        valid_indices = []
        
        for idx, row in self.data.iterrows():
            smiles = row['smiles']
            mol = self._process_molecule(smiles)
            
            if mol is not None:
                desc = {}
                
                # Basic descriptors (2D)
                for name, func in Descriptors.descList:
                    try:
                        desc[name] = func(mol)
                    except:
                        desc[name] = np.nan
                
                # Additional 2D descriptors
                desc['NumAtoms'] = mol.GetNumAtoms()
                desc['NumHeavyAtoms'] = mol.GetNumHeavyAtoms()
                desc['NumRotatableBonds'] = rdMolDescriptors.CalcNumRotatableBonds(mol)
                desc['NumHDonors'] = rdMolDescriptors.CalcNumHBD(mol)
                desc['NumHAcceptors'] = rdMolDescriptors.CalcNumHBA(mol)
                desc['TPSA'] = rdMolDescriptors.CalcTPSA(mol)
                desc['MolLogP'] = Descriptors.MolLogP(mol)
                desc['MolWt'] = Descriptors.MolWt(mol)
                desc['RingCount'] = rdMolDescriptors.CalcNumRings(mol)
                
                # 3D descriptors
                try:
                    desc['PMI1'] = rdMolDescriptors.CalcPMI1(mol)
                    desc['PMI2'] = rdMolDescriptors.CalcPMI2(mol)
                    desc['PMI3'] = rdMolDescriptors.CalcPMI3(mol)
                    desc['NPR1'] = rdMolDescriptors.CalcNPR1(mol)
                    desc['NPR2'] = rdMolDescriptors.CalcNPR2(mol)
                    desc['RadiusOfGyration'] = rdMolDescriptors.CalcRadiusOfGyration(mol)
                    desc['InertialShapeFactor'] = rdMolDescriptors.CalcInertialShapeFactor(mol)
                    desc['Eccentricity'] = rdMolDescriptors.CalcEccentricity(mol)
                    desc['Asphericity'] = rdMolDescriptors.CalcAsphericity(mol)
                    desc['SpherocityIndex'] = rdMolDescriptors.CalcSpherocityIndex(mol)
                except:
                    pass
                
                descriptor_list.append(desc)
                temp_list.append(self.temp_data.loc[idx].values)
                valid_indices.append(idx)
            else:
                print(f"Could not process SMILES: {smiles}")
        
        # Create descriptor and temperature DataFrames
        self.descriptors = pd.DataFrame(descriptor_list, index=valid_indices)
        self.temp_data = pd.DataFrame(temp_list, columns=['Inverse_Temperature_K'], index=valid_indices)
        self.data = self.data.loc[valid_indices]
        
        print(f"Generated {self.descriptors.shape[1]} molecular descriptors + 1 temperature feature for {len(self.data)} valid compounds")
        return self.descriptors
    
    def preprocess_data(self, test_size=0.2, random_state=42, strategy='mean', variance_threshold=0.001):
        """Preprocess data combining descriptors and temperature features with variance filtering"""
        if self.descriptors is None:
            raise ValueError("No descriptors generated. Please generate descriptors first.")
            
        # Combine RDKit descriptors with Inverse Temperature feature
        X_combined = pd.concat([self.descriptors, self.temp_data], axis=1)
        
        # Remove constant features
        selector = VarianceThreshold(threshold=variance_threshold)
        selected_features = selector.fit_transform(X_combined)
        
        # Get feature names (after filtering)
        self.feature_names = np.array(X_combined.columns)[selector.get_support()]
        
        # Ensure Inverse_Temperature_K is preserved even if variance is low
        if 'Inverse_Temperature_K' not in self.feature_names:
            X_combined_filtered = pd.DataFrame(selected_features, columns=self.feature_names)
            X_combined_filtered['Inverse_Temperature_K'] = X_combined['Inverse_Temperature_K'].values
            self.feature_names = list(self.feature_names) + ['Inverse_Temperature_K']
            X_filtered = X_combined_filtered
        else:
            X_filtered = pd.DataFrame(selected_features, columns=self.feature_names)
        
        # Split data
        self.X = X_filtered
        self.y = self.data['property']
        
        # Split into train/test
        self.X_train, self.X_test, self.y_train, self.y_test = train_test_split(
            self.X, self.y, test_size=test_size, random_state=random_state)
        
        # Create imputer and scaler
        self.imputer = SimpleImputer(strategy=strategy)
        self.scaler = StandardScaler()
        
        # Fit on training data only and transform
        self.X_train = pd.DataFrame(
            self.scaler.fit_transform(self.imputer.fit_transform(self.X_train)),
            columns=self.feature_names
        )
        
        # Transform test data
        self.X_test = pd.DataFrame(
            self.scaler.transform(self.imputer.transform(self.X_test)),
            columns=self.feature_names
        )
        
        print(f"Data preprocessed. Train size: {len(self.y_train)}, Test size: {len(self.y_test)}")
        print(f"Features after variance threshold: {len(self.feature_names)}")
    
    def hyperparameter_tuning(self, search_type='grid', cv=5, n_iter=20):
        """Perform hyperparameter optimization for Neural Network"""
        print("\n=== Neural Network Hyperparameter Optimization ===")
        
        param_grid = {
            'mlpregressor__hidden_layer_sizes': [
                (50,), (100,), (50, 25), (100, 50), (100, 50, 25),
                (150, 75, 30), (200, 100, 50), (50, 100, 50)
            ],
            'mlpregressor__activation': ['relu', 'tanh'],
            'mlpregressor__alpha': [0.0001, 0.001, 0.01, 0.1],
            'mlpregressor__learning_rate': ['constant', 'adaptive'],
            'mlpregressor__learning_rate_init': [0.0001, 0.001, 0.01],
            'mlpregressor__solver': ['adam'],
            'mlpregressor__max_iter': [1000, 1500]
        }
        
        base_nn = MLPRegressor(random_state=42)
        
        pipeline = make_pipeline(
            SimpleImputer(strategy='mean'),
            StandardScaler(),
            base_nn
        )
        
        if search_type == 'grid':
            print("Performing Grid Search...")
            search = GridSearchCV(
                pipeline, param_grid, cv=cv, scoring='r2', n_jobs=-1, verbose=1
            )
        else:
            print(f"Performing Randomized Search with {n_iter} iterations...")
            search = RandomizedSearchCV(
                pipeline, param_grid, n_iter=n_iter, cv=cv, scoring='r2', n_jobs=-1, verbose=1, random_state=42
            )
        
        search.fit(self.X_train, self.y_train)
        
        self.best_model = search.best_estimator_
        self.best_params = search.best_params_
        self.best_r2 = search.best_score_
        self.best_model_name = 'Neural Network (Optimized)'
        
        print(f"\nBest parameters found:")
        for param, value in self.best_params.items():
            print(f"  {param}: {value}")
        print(f"\nBest CV R² score: {self.best_r2:.3f}")
        
        test_pred = self.best_model.predict(self.X_test)
        test_r2 = r2_score(self.y_test, test_pred)
        print(f"Test R² score: {test_r2:.3f}")
        
        self.save_best_model()
        return search
    
    def train_simple_model(self, hidden_layer_sizes=(100, 50), activation='relu', alpha=0.001, 
                           learning_rate='adaptive', max_iter=1000):
        """Train a simple neural network without hyperparameter tuning"""
        print("\n=== Training Simple Neural Network ===")
        
        nn = MLPRegressor(
            hidden_layer_sizes=hidden_layer_sizes,
            activation=activation,
            alpha=alpha,
            learning_rate=learning_rate,
            max_iter=max_iter,
            random_state=42
        )
        
        self.model = make_pipeline(
            SimpleImputer(strategy='mean'),
            StandardScaler(),
            nn
        )
        
        self.model.fit(self.X_train, self.y_train)
        self.best_model = self.model
        self.best_model_name = 'Neural Network'
        
        test_pred = self.model.predict(self.X_test)
        self.best_r2 = r2_score(self.y_test, test_pred)
        
        self.save_best_model()
        return self.model
    
    def cross_validate(self, cv=5):
        """Perform cross-validation"""
        if self.best_model is None:
            raise ValueError("No model trained. Please train a model first.")
            
        print("\n=== Cross-Validation ===")
        r2_scores = cross_val_score(self.best_model, self.X_train, self.y_train, cv=cv, scoring='r2')
        print(f"R² scores: {r2_scores}")
        print(f"Mean R²: {np.mean(r2_scores):.4f} ± {np.std(r2_scores):.4f}")
    
    def save_best_model(self):
        """Save the best performing model to disk"""
        if self.best_model is None:
            raise ValueError("No best model identified yet")
            
        filename = f"best_model_{self.best_model_name.replace(' ', '_')}.pkl"
        filepath = os.path.join(self.output_dir, filename)
        
        model_info = {
            'model': self.best_model,
            'scaler': self.scaler,
            'imputer': self.imputer,
            'feature_names': self.feature_names,
            'model_name': self.best_model_name,
            'test_r2': self.best_r2,
            'best_params': self.best_params
        }
        
        joblib.dump(model_info, filepath)
        print(f"\nSaved best model ({self.best_model_name}) to {filepath}")
    
    def plot_parity(self, model=None):
        """Create and save parity plot for best model"""
        if model is None:
            if self.best_model is None:
                raise ValueError("No best model available")
            model = self.best_model
            model_name = self.best_model_name
        else:
            model_name = "Selected Model"
        
        train_pred = model.predict(self.X_train)
        test_pred = model.predict(self.X_test)
        
        plt.figure(figsize=(8, 6))
        plt.scatter(self.y_train, train_pred, alpha=1, label='Train')
        plt.scatter(self.y_test, test_pred, alpha=1, label='Test')
        
        min_val = min(min(self.y_train), min(self.y_test))
        max_val = max(max(self.y_train), max(self.y_test))
        plt.plot([min_val, max_val], [min_val, max_val], 'k--')
        
        plt.xlabel('Experimental Value')
        plt.ylabel('Predicted Value')
        plt.title(f'Parity Plot - {model_name} (with Temperature)')
        plt.legend()
        plt.grid(True)
        
        plt.savefig(os.path.join(self.output_dir, 'parity_plot.png'), dpi=300, bbox_inches='tight')
        plt.show()
    
    def plot_loss_curve(self):
        """Plot training loss curve for neural network"""
        if self.best_model is None:
            print("No model available")
            return
            
        nn = self.best_model.named_steps['mlpregressor']
        if hasattr(nn, 'loss_curve_'):
            plt.figure(figsize=(8, 5))
            plt.plot(nn.loss_curve_)
            plt.title('Training Loss Curve - Neural Network')
            plt.xlabel('Epochs')
            plt.ylabel('Loss')
            plt.grid(True)
            
            plt.savefig(os.path.join(self.output_dir, 'loss_curve.png'), dpi=300, bbox_inches='tight')
            plt.show()
        else:
            print("Loss curve not available for this model")
    
    def plot_feature_importance(self, top_n=20):
        """Plot feature importance based on neural network weights with adjusted margins to prevent text clipping"""
        if self.best_model is None:
            raise ValueError("No best model available")
        
        nn = self.best_model.named_steps['mlpregressor']
        
        if hasattr(nn, 'coefs_'):
            importance = np.abs(nn.coefs_[0]).sum(axis=1)
            
            idx = np.argsort(importance)[::-1]
            sorted_names = [self.feature_names[i] for i in idx]
            sorted_importance = importance[idx]
            
            plt.figure(figsize=(10, 6))
            plt.barh(range(min(top_n, len(sorted_importance))), 
                     sorted_importance[:top_n][::-1], 
                     align='center')
            plt.yticks(range(min(top_n, len(sorted_importance))), 
                       sorted_names[:top_n][::-1])
            plt.xlabel('Feature Importance (Sum of Absolute Weights)')
            plt.title(f'Top {top_n} Features - Neural Network')
            
            # Add extra space on the left side to avoid cutting off long feature names (like Inverse_Temperature_K)
            plt.subplots_adjust(left=0.3)
            plt.tight_layout()
            
            plt.savefig(os.path.join(self.output_dir, 'feature_importance.png'), dpi=300, bbox_inches='tight')
            plt.show()
        else:
            print("Feature weights not available for this model")

# Example usage
if __name__ == "__main__":
    predictor = Chemical_Property_Predictor_NN(output_dir='best_models')
    
    # Load data using your file name format
    predictor.load_data('Original_data.xlsx')
    
    # Generate descriptors and include temperature
    predictor.generate_descriptors()
    
    # Preprocess data
    predictor.preprocess_data()
    
    # Run hyperparameter optimization or training
    predictor.hyperparameter_tuning(search_type='randomized', cv=5, n_iter=10)
    
    
    # Cross-validation
    predictor.cross_validate()
    
    # Generate visualizations & save outputs automatically
    predictor.plot_parity()
    predictor.plot_loss_curve()
    predictor.plot_feature_importance()
```

```python
# -*- coding: utf-8 -*-
"""
Created on Sun Jun 28 15:03:58 2026

@author: user
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from rdkit import Chem
from rdkit.Chem import AllChem, Descriptors, rdMolDescriptors
from rdkit.Chem.MolStandardize import rdMolStandardize
from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV, RandomizedSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
from sklearn.feature_selection import VarianceThreshold
from sklearn.impute import SimpleImputer
from sklearn.pipeline import make_pipeline
from sklearn.neural_network import MLPRegressor
import joblib
import os
import warnings
warnings.filterwarnings('ignore')

class Chemical_Property_Predictor_NN:
    def __init__(self, output_dir='best_models'):
        self.raw_data = None
        self.data = None
        self.temp_data = None
        self.descriptors = None
        self.X = None
        self.y = None
        self.X_train = None
        self.X_test = None
        self.y_train = None
        self.y_test = None
        self.scaler = None
        self.imputer = None
        self.model = None
        self.best_model = None
        self.best_model_name = None
        self.best_r2 = -np.inf
        self.best_params = None
        self.output_dir = output_dir
        self.feature_names = None
        
        # Create output directory if it doesn't exist
        os.makedirs(self.output_dir, exist_ok=True)
        
        # Initialize standardizer for salt stripping
        self.normalizer = rdMolStandardize.Normalizer()
        self.uncharger = rdMolStandardize.Uncharger()
        
    def _process_molecule(self, smiles):
        """Process a single molecule through all steps"""
        try:
            # 1. Read SMILES
            mol = Chem.MolFromSmiles(smiles)
            if mol is None:
                return None
            
            # 2. Transform to molecule and sanitize
            mol = Chem.MolFromSmiles(Chem.MolToSmiles(mol))
            if mol is None:
                return None
            
            # 3. Strip salts and standardize
            parent_mol = rdMolStandardize.FragmentParent(mol)
            
            # 4. Add hydrogens
            mol = Chem.AddHs(parent_mol)
            
            # 5. Generate coordinates
            try:
                AllChem.EmbedMolecule(mol, randomSeed=42)
                AllChem.MMFFOptimizeMolecule(mol)
            except:
                pass
            
            return mol
            
        except Exception as e:
            print(f"Error processing molecule {smiles}: {str(e)}")
            return None
    
    def load_data(self, file_path):
        """Load chemical data from CSV or Excel file, handling name mappings and Inverse_Temperature_K"""
        if file_path.endswith(('.xlsx', '.xls')):
            self.raw_data = pd.read_excel(file_path)
        else:
            self.raw_data = pd.read_csv(file_path)
        
        rename_map = {}
        if 'NAME' in self.raw_data.columns:
            rename_map['NAME'] = 'substance_name'
        if 'SMILES' in self.raw_data.columns:
            rename_map['SMILES'] = 'smiles'
        if 'Log_Psat_mmHg' in self.raw_data.columns:
            rename_map['Log_Psat_mmHg'] = 'property'
            
        if rename_map:
            self.raw_data = self.raw_data.rename(columns=rename_map)
            
        required = ['substance_name', 'smiles', 'property', 'Inverse_Temperature_K']
        for col in required:
            if col not in self.raw_data.columns:
                raise ValueError(f"Column '{col}' not found. Available columns: {list(self.raw_data.columns)}")
        
        self.data = self.raw_data[['substance_name', 'smiles', 'property']].copy()
        self.temp_data = self.raw_data[['Inverse_Temperature_K']].copy()
        print(f"Loaded {len(self.data)} rows from {file_path}")

    def generate_descriptors(self):
        """Generate molecular descriptors using RDKit with full processing and retain Inverse_Temperature_K"""
        if self.data is None:
            raise ValueError("No data loaded. Please load data first.")
            
        descriptor_list = []
        temp_list = []
        valid_indices = []
        
        for idx, row in self.data.iterrows():
            smiles = row['smiles']
            mol = self._process_molecule(smiles)
            
            if mol is not None:
                desc = {}
                
                # Basic descriptors (2D)
                for name, func in Descriptors.descList:
                    try:
                        desc[name] = func(mol)
                    except:
                        desc[name] = np.nan
                
                # Additional 2D descriptors
                desc['NumAtoms'] = mol.GetNumAtoms()
                desc['NumHeavyAtoms'] = mol.GetNumHeavyAtoms()
                desc['NumRotatableBonds'] = rdMolDescriptors.CalcNumRotatableBonds(mol)
                desc['NumHDonors'] = rdMolDescriptors.CalcNumHBD(mol)
                desc['NumHAcceptors'] = rdMolDescriptors.CalcNumHBA(mol)
                desc['TPSA'] = rdMolDescriptors.CalcTPSA(mol)
                desc['MolLogP'] = Descriptors.MolLogP(mol)
                desc['MolWt'] = Descriptors.MolWt(mol)
                desc['RingCount'] = rdMolDescriptors.CalcNumRings(mol)
                
                # 3D descriptors
                try:
                    desc['PMI1'] = rdMolDescriptors.CalcPMI1(mol)
                    desc['PMI2'] = rdMolDescriptors.CalcPMI2(mol)
                    desc['PMI3'] = rdMolDescriptors.CalcPMI3(mol)
                    desc['NPR1'] = rdMolDescriptors.CalcNPR1(mol)
                    desc['NPR2'] = rdMolDescriptors.CalcNPR2(mol)
                    desc['RadiusOfGyration'] = rdMolDescriptors.CalcRadiusOfGyration(mol)
                    desc['InertialShapeFactor'] = rdMolDescriptors.CalcInertialShapeFactor(mol)
                    desc['Eccentricity'] = rdMolDescriptors.CalcEccentricity(mol)
                    desc['Asphericity'] = rdMolDescriptors.CalcAsphericity(mol)
                    desc['SpherocityIndex'] = rdMolDescriptors.CalcSpherocityIndex(mol)
                except:
                    pass
                
                descriptor_list.append(desc)
                temp_list.append(self.temp_data.loc[idx].values)
                valid_indices.append(idx)
            else:
                print(f"Could not process SMILES: {smiles}")
        
        # Create descriptor and temperature DataFrames
        self.descriptors = pd.DataFrame(descriptor_list, index=valid_indices)
        self.temp_data = pd.DataFrame(temp_list, columns=['Inverse_Temperature_K'], index=valid_indices)
        self.data = self.data.loc[valid_indices]
        
        print(f"Generated {self.descriptors.shape[1]} molecular descriptors + 1 temperature feature for {len(self.data)} valid compounds")
        return self.descriptors
    
    def preprocess_data(self, test_size=0.2, random_state=42, strategy='mean', variance_threshold=0.001):
        """Preprocess data combining descriptors and temperature features with variance filtering"""
        if self.descriptors is None:
            raise ValueError("No descriptors generated. Please generate descriptors first.")
            
        # Combine RDKit descriptors with Inverse Temperature feature
        X_combined = pd.concat([self.descriptors, self.temp_data], axis=1)
        
        # Remove constant features
        selector = VarianceThreshold(threshold=variance_threshold)
        selected_features = selector.fit_transform(X_combined)
        
        # Get feature names (after filtering)
        self.feature_names = np.array(X_combined.columns)[selector.get_support()]
        
        # Ensure Inverse_Temperature_K is preserved even if variance is low
        if 'Inverse_Temperature_K' not in self.feature_names:
            X_combined_filtered = pd.DataFrame(selected_features, columns=self.feature_names)
            X_combined_filtered['Inverse_Temperature_K'] = X_combined['Inverse_Temperature_K'].values
            self.feature_names = list(self.feature_names) + ['Inverse_Temperature_K']
            X_filtered = X_combined_filtered
        else:
            X_filtered = pd.DataFrame(selected_features, columns=self.feature_names)
        
        # Split data
        self.X = X_filtered
        self.y = self.data['property']
        
        # Split into train/test
        self.X_train, self.X_test, self.y_train, self.y_test = train_test_split(
            self.X, self.y, test_size=test_size, random_state=random_state)
        
        # Create imputer and scaler
        self.imputer = SimpleImputer(strategy=strategy)
        self.scaler = StandardScaler()
        
        # Fit on training data only and transform
        self.X_train = pd.DataFrame(
            self.scaler.fit_transform(self.imputer.fit_transform(self.X_train)),
            columns=self.feature_names
        )
        
        # Transform test data
        self.X_test = pd.DataFrame(
            self.scaler.transform(self.imputer.transform(self.X_test)),
            columns=self.feature_names
        )
        
        print(f"Data preprocessed. Train size: {len(self.y_train)}, Test size: {len(self.y_test)}")
        print(f"Features after variance threshold: {len(self.feature_names)}")
    
    def hyperparameter_tuning(self, search_type='grid', cv=5, n_iter=20):
        """Perform hyperparameter optimization for Neural Network"""
        print("\n=== Neural Network Hyperparameter Optimization ===")
        
        param_grid = {
            'mlpregressor__hidden_layer_sizes': [
                (50,), (100,), (50, 25), (100, 50), (100, 50, 25),
                (150, 75, 30), (200, 100, 50), (50, 100, 50)
            ],
            'mlpregressor__activation': ['relu', 'tanh'],
            'mlpregressor__alpha': [0.0001, 0.001, 0.01, 0.1],
            'mlpregressor__learning_rate': ['constant', 'adaptive'],
            'mlpregressor__learning_rate_init': [0.0001, 0.001, 0.01],
            'mlpregressor__solver': ['adam'],
            'mlpregressor__max_iter': [1000, 1500]
        }
        
        base_nn = MLPRegressor(random_state=42)
        
        pipeline = make_pipeline(
            SimpleImputer(strategy='mean'),
            StandardScaler(),
            base_nn
        )
        
        if search_type == 'grid':
            print("Performing Grid Search...")
            search = GridSearchCV(
                pipeline, param_grid, cv=cv, scoring='r2', n_jobs=-1, verbose=1
            )
        else:
            print(f"Performing Randomized Search with {n_iter} iterations...")
            search = RandomizedSearchCV(
                pipeline, param_grid, n_iter=n_iter, cv=cv, scoring='r2', n_jobs=-1, verbose=1, random_state=42
            )
        
        search.fit(self.X_train, self.y_train)
        
        self.best_model = search.best_estimator_
        self.best_params = search.best_params_
        self.best_r2 = search.best_score_
        self.best_model_name = 'Neural Network (Optimized)'
        
        print(f"\nBest parameters found:")
        for param, value in self.best_params.items():
            print(f"  {param}: {value}")
        print(f"\nBest CV R² score: {self.best_r2:.3f}")
        
        test_pred = self.best_model.predict(self.X_test)
        test_r2 = r2_score(self.y_test, test_pred)
        print(f"Test R² score: {test_r2:.3f}")
        
        self.save_best_model()
        return search
    
    def train_simple_model(self, hidden_layer_sizes=(100, 50), activation='relu', alpha=0.001, 
                           learning_rate='adaptive', max_iter=1000):
        """Train a simple neural network without hyperparameter tuning"""
        print("\n=== Training Simple Neural Network ===")
        
        nn = MLPRegressor(
            hidden_layer_sizes=hidden_layer_sizes,
            activation=activation,
            alpha=alpha,
            learning_rate=learning_rate,
            max_iter=max_iter,
            random_state=42
        )
        
        self.model = make_pipeline(
            SimpleImputer(strategy='mean'),
            StandardScaler(),
            nn
        )
        
        self.model.fit(self.X_train, self.y_train)
        self.best_model = self.model
        self.best_model_name = 'Neural Network'
        
        test_pred = self.model.predict(self.X_test)
        self.best_r2 = r2_score(self.y_test, test_pred)
        
        self.save_best_model()
        return self.model
    
    def cross_validate(self, cv=5):
        """Perform cross-validation"""
        if self.best_model is None:
            raise ValueError("No model trained. Please train a model first.")
            
        print("\n=== Cross-Validation ===")
        r2_scores = cross_val_score(self.best_model, self.X_train, self.y_train, cv=cv, scoring='r2')
        print(f"R² scores: {r2_scores}")
        print(f"Mean R²: {np.mean(r2_scores):.4f} ± {np.std(r2_scores):.4f}")
    
    def save_best_model(self):
        """Save the best performing model to disk"""
        if self.best_model is None:
            raise ValueError("No best model identified yet")
            
        filename = f"best_model_{self.best_model_name.replace(' ', '_')}.pkl"
        filepath = os.path.join(self.output_dir, filename)
        
        model_info = {
            'model': self.best_model,
            'scaler': self.scaler,
            'imputer': self.imputer,
            'feature_names': self.feature_names,
            'model_name': self.best_model_name,
            'test_r2': self.best_r2,
            'best_params': self.best_params
        }
        
        joblib.dump(model_info, filepath)
        print(f"\nSaved best model ({self.best_model_name}) to {filepath}")
    
    def plot_parity(self, model=None):
        """Create and save parity plot for best model"""
        if model is None:
            if self.best_model is None:
                raise ValueError("No best model available")
            model = self.best_model
            model_name = self.best_model_name
        else:
            model_name = "Selected Model"
        
        train_pred = model.predict(self.X_train)
        test_pred = model.predict(self.X_test)
        
        plt.figure(figsize=(8, 6))
        plt.scatter(self.y_train, train_pred, alpha=1, label='Train')
        plt.scatter(self.y_test, test_pred, alpha=1, label='Test')
        
        min_val = min(min(self.y_train), min(self.y_test))
        max_val = max(max(self.y_train), max(self.y_test))
        plt.plot([min_val, max_val], [min_val, max_val], 'k--')
        
        plt.xlabel('Experimental Value')
        plt.ylabel('Predicted Value')
        plt.title(f'Parity Plot - {model_name} (with Temperature)')
        plt.legend()
        plt.grid(True)
        
        plt.savefig(os.path.join(self.output_dir, 'parity_plot.png'), dpi=300, bbox_inches='tight')
        plt.show()
    
    def plot_loss_curve(self):
        """Plot training loss curve for neural network"""
        if self.best_model is None:
            print("No model available")
            return
            
        nn = self.best_model.named_steps['mlpregressor']
        if hasattr(nn, 'loss_curve_'):
            plt.figure(figsize=(8, 5))
            plt.plot(nn.loss_curve_)
            plt.title('Training Loss Curve - Neural Network')
            plt.xlabel('Epochs')
            plt.ylabel('Loss')
            plt.grid(True)
            
            plt.savefig(os.path.join(self.output_dir, 'loss_curve.png'), dpi=300, bbox_inches='tight')
            plt.show()
        else:
            print("Loss curve not available for this model")
    
    def plot_feature_importance(self, top_n=20):
        """Plot feature importance based on neural network weights with adjusted margins to prevent text clipping"""
        if self.best_model is None:
            raise ValueError("No best model available")
        
        nn = self.best_model.named_steps['mlpregressor']
        
        if hasattr(nn, 'coefs_'):
            importance = np.abs(nn.coefs_[0]).sum(axis=1)
            
            idx = np.argsort(importance)[::-1]
            sorted_names = [self.feature_names[i] for i in idx]
            sorted_importance = importance[idx]
            
            plt.figure(figsize=(10, 6))
            plt.barh(range(min(top_n, len(sorted_importance))), 
                     sorted_importance[:top_n][::-1], 
                     align='center')
            plt.yticks(range(min(top_n, len(sorted_importance))), 
                       sorted_names[:top_n][::-1])
            plt.xlabel('Feature Importance (Sum of Absolute Weights)')
            plt.title(f'Top {top_n} Features - Neural Network')
            
            # Add extra space on the left side to avoid cutting off long feature names (like Inverse_Temperature_K)
            plt.subplots_adjust(left=0.3)
            plt.tight_layout()
            
            plt.savefig(os.path.join(self.output_dir, 'feature_importance.png'), dpi=300, bbox_inches='tight')
            plt.show()
        else:
            print("Feature weights not available for this model")

# Example usage
if __name__ == "__main__":
    predictor = Chemical_Property_Predictor_NN(output_dir='best_models')
    
    # Load data using your file name format
    predictor.load_data('Original_data_filtered.xlsx')
    
    # Generate descriptors and include temperature
    predictor.generate_descriptors()
    
    # Preprocess data
    predictor.preprocess_data()
    
    # Run hyperparameter optimization or training
    predictor.hyperparameter_tuning(search_type='randomized', cv=5, n_iter=10)
    
    
    # Cross-validation
    predictor.cross_validate()
    
    # Generate visualizations & save outputs automatically
    predictor.plot_parity()
    predictor.plot_loss_curve()
    predictor.plot_feature_importance()
```

```python
# -*- coding: utf-8 -*-
"""
Created on Sun Jun 28 15:03:58 2026

@author: user
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from rdkit import Chem
from rdkit.Chem import AllChem, Descriptors, rdMolDescriptors
from rdkit.Chem.MolStandardize import rdMolStandardize
from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV, RandomizedSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
from sklearn.feature_selection import VarianceThreshold
from sklearn.impute import SimpleImputer
from sklearn.pipeline import make_pipeline
from sklearn.neural_network import MLPRegressor
import joblib
import os
import warnings
warnings.filterwarnings('ignore')

class Chemical_Property_Predictor_NN:
    def __init__(self, output_dir='best_models'):
        self.raw_data = None
        self.data = None
        self.temp_data = None
        self.descriptors = None
        self.X = None
        self.y = None
        self.X_train = None
        self.X_test = None
        self.y_train = None
        self.y_test = None
        self.scaler = None
        self.imputer = None
        self.model = None
        self.best_model = None
        self.best_model_name = None
        self.best_r2 = -np.inf
        self.best_params = None
        self.output_dir = output_dir
        self.feature_names = None
        
        # Create output directory if it doesn't exist
        os.makedirs(self.output_dir, exist_ok=True)
        
        # Initialize standardizer for salt stripping
        self.normalizer = rdMolStandardize.Normalizer()
        self.uncharger = rdMolStandardize.Uncharger()
        
    def _process_molecule(self, smiles):
        """Process a single molecule through all steps"""
        try:
            # 1. Read SMILES
            mol = Chem.MolFromSmiles(smiles)
            if mol is None:
                return None
            
            # 2. Transform to molecule and sanitize
            mol = Chem.MolFromSmiles(Chem.MolToSmiles(mol))
            if mol is None:
                return None
            
            # 3. Strip salts and standardize
            parent_mol = rdMolStandardize.FragmentParent(mol)
            
            # 4. Add hydrogens
            mol = Chem.AddHs(parent_mol)
            
            # 5. Generate coordinates
            try:
                AllChem.EmbedMolecule(mol, randomSeed=42)
                AllChem.MMFFOptimizeMolecule(mol)
            except:
                pass
            
            return mol
            
        except Exception as e:
            print(f"Error processing molecule {smiles}: {str(e)}")
            return None
    
    def load_data(self, file_path):
        """Load chemical data from CSV or Excel file, handling name mappings and Inverse_Temperature_K"""
        if file_path.endswith(('.xlsx', '.xls')):
            self.raw_data = pd.read_excel(file_path)
        else:
            self.raw_data = pd.read_csv(file_path)
        
        rename_map = {}
        if 'NAME' in self.raw_data.columns:
            rename_map['NAME'] = 'substance_name'
        if 'SMILES' in self.raw_data.columns:
            rename_map['SMILES'] = 'smiles'
        if 'Log_Psat_mmHg' in self.raw_data.columns:
            rename_map['Log_Psat_mmHg'] = 'property'
            
        if rename_map:
            self.raw_data = self.raw_data.rename(columns=rename_map)
            
        required = ['substance_name', 'smiles', 'property', 'Inverse_Temperature_K']
        for col in required:
            if col not in self.raw_data.columns:
                raise ValueError(f"Column '{col}' not found. Available columns: {list(self.raw_data.columns)}")
        
        self.data = self.raw_data[['substance_name', 'smiles', 'property']].copy()
        self.temp_data = self.raw_data[['Inverse_Temperature_K']].copy()
        print(f"Loaded {len(self.data)} rows from {file_path}")

    def generate_descriptors(self):
        """Generate molecular descriptors using RDKit with full processing and retain Inverse_Temperature_K"""
        if self.data is None:
            raise ValueError("No data loaded. Please load data first.")
            
        descriptor_list = []
        temp_list = []
        valid_indices = []
        
        for idx, row in self.data.iterrows():
            smiles = row['smiles']
            mol = self._process_molecule(smiles)
            
            if mol is not None:
                desc = {}
                
                # Basic descriptors (2D)
                for name, func in Descriptors.descList:
                    try:
                        desc[name] = func(mol)
                    except:
                        desc[name] = np.nan
                
                # Additional 2D descriptors
                desc['NumAtoms'] = mol.GetNumAtoms()
                desc['NumHeavyAtoms'] = mol.GetNumHeavyAtoms()
                desc['NumRotatableBonds'] = rdMolDescriptors.CalcNumRotatableBonds(mol)
                desc['NumHDonors'] = rdMolDescriptors.CalcNumHBD(mol)
                desc['NumHAcceptors'] = rdMolDescriptors.CalcNumHBA(mol)
                desc['TPSA'] = rdMolDescriptors.CalcTPSA(mol)
                desc['MolLogP'] = Descriptors.MolLogP(mol)
                desc['MolWt'] = Descriptors.MolWt(mol)
                desc['RingCount'] = rdMolDescriptors.CalcNumRings(mol)
                
                # 3D descriptors
                try:
                    desc['PMI1'] = rdMolDescriptors.CalcPMI1(mol)
                    desc['PMI2'] = rdMolDescriptors.CalcPMI2(mol)
                    desc['PMI3'] = rdMolDescriptors.CalcPMI3(mol)
                    desc['NPR1'] = rdMolDescriptors.CalcNPR1(mol)
                    desc['NPR2'] = rdMolDescriptors.CalcNPR2(mol)
                    desc['RadiusOfGyration'] = rdMolDescriptors.CalcRadiusOfGyration(mol)
                    desc['InertialShapeFactor'] = rdMolDescriptors.CalcInertialShapeFactor(mol)
                    desc['Eccentricity'] = rdMolDescriptors.CalcEccentricity(mol)
                    desc['Asphericity'] = rdMolDescriptors.CalcAsphericity(mol)
                    desc['SpherocityIndex'] = rdMolDescriptors.CalcSpherocityIndex(mol)
                except:
                    pass
                
                descriptor_list.append(desc)
                temp_list.append(self.temp_data.loc[idx].values)
                valid_indices.append(idx)
            else:
                print(f"Could not process SMILES: {smiles}")
        
        # Create descriptor and temperature DataFrames
        self.descriptors = pd.DataFrame(descriptor_list, index=valid_indices)
        self.temp_data = pd.DataFrame(temp_list, columns=['Inverse_Temperature_K'], index=valid_indices)
        self.data = self.data.loc[valid_indices]
        
        print(f"Generated {self.descriptors.shape[1]} molecular descriptors + 1 temperature feature for {len(self.data)} valid compounds")
        return self.descriptors
    
    def preprocess_data(self, test_size=0.2, random_state=42, strategy='mean', variance_threshold=0.001):
        """Preprocess data combining descriptors and temperature features with variance filtering"""
        if self.descriptors is None:
            raise ValueError("No descriptors generated. Please generate descriptors first.")
            
        # Combine RDKit descriptors with Inverse Temperature feature
        X_combined = pd.concat([self.descriptors, self.temp_data], axis=1)
        
        # Remove constant features
        selector = VarianceThreshold(threshold=variance_threshold)
        selected_features = selector.fit_transform(X_combined)
        
        # Get feature names (after filtering)
        self.feature_names = np.array(X_combined.columns)[selector.get_support()]
        
        # Ensure Inverse_Temperature_K is preserved even if variance is low
        if 'Inverse_Temperature_K' not in self.feature_names:
            X_combined_filtered = pd.DataFrame(selected_features, columns=self.feature_names)
            X_combined_filtered['Inverse_Temperature_K'] = X_combined['Inverse_Temperature_K'].values
            self.feature_names = list(self.feature_names) + ['Inverse_Temperature_K']
            X_filtered = X_combined_filtered
        else:
            X_filtered = pd.DataFrame(selected_features, columns=self.feature_names)
        
        # Split data
        self.X = X_filtered
        self.y = self.data['property']
        
        # Split into train/test
        self.X_train, self.X_test, self.y_train, self.y_test = train_test_split(
            self.X, self.y, test_size=test_size, random_state=random_state)
        
        # Create imputer and scaler
        self.imputer = SimpleImputer(strategy=strategy)
        self.scaler = StandardScaler()
        
        # Fit on training data only and transform
        self.X_train = pd.DataFrame(
            self.scaler.fit_transform(self.imputer.fit_transform(self.X_train)),
            columns=self.feature_names
        )
        
        # Transform test data
        self.X_test = pd.DataFrame(
            self.scaler.transform(self.imputer.transform(self.X_test)),
            columns=self.feature_names
        )
        
        print(f"Data preprocessed. Train size: {len(self.y_train)}, Test size: {len(self.y_test)}")
        print(f"Features after variance threshold: {len(self.feature_names)}")
    
    def hyperparameter_tuning(self, search_type='grid', cv=5, n_iter=20):
        """Perform hyperparameter optimization for Neural Network"""
        print("\n=== Neural Network Hyperparameter Optimization ===")
        
        param_grid = {
            'mlpregressor__hidden_layer_sizes': [
                (50,), (100,), (50, 25), (100, 50), (100, 50, 25),
                (150, 75, 30), (200, 100, 50), (50, 100, 50)
            ],
            'mlpregressor__activation': ['relu', 'tanh'],
            'mlpregressor__alpha': [0.0001, 0.001, 0.01, 0.1],
            'mlpregressor__learning_rate': ['constant', 'adaptive'],
            'mlpregressor__learning_rate_init': [0.0001, 0.001, 0.01],
            'mlpregressor__solver': ['adam'],
            'mlpregressor__max_iter': [1000, 1500]
        }
        
        base_nn = MLPRegressor(random_state=42)
        
        pipeline = make_pipeline(
            SimpleImputer(strategy='mean'),
            StandardScaler(),
            base_nn
        )
        
        if search_type == 'grid':
            print("Performing Grid Search...")
            search = GridSearchCV(
                pipeline, param_grid, cv=cv, scoring='r2', n_jobs=-1, verbose=1
            )
        else:
            print(f"Performing Randomized Search with {n_iter} iterations...")
            search = RandomizedSearchCV(
                pipeline, param_grid, n_iter=n_iter, cv=cv, scoring='r2', n_jobs=-1, verbose=1, random_state=42
            )
        
        search.fit(self.X_train, self.y_train)
        
        self.best_model = search.best_estimator_
        self.best_params = search.best_params_
        self.best_r2 = search.best_score_
        self.best_model_name = 'Neural Network (Optimized)'
        
        print(f"\nBest parameters found:")
        for param, value in self.best_params.items():
            print(f"  {param}: {value}")
        print(f"\nBest CV R² score: {self.best_r2:.3f}")
        
        test_pred = self.best_model.predict(self.X_test)
        test_r2 = r2_score(self.y_test, test_pred)
        print(f"Test R² score: {test_r2:.3f}")
        
        self.save_best_model()
        return search
    
    def train_simple_model(self, hidden_layer_sizes=(100, 50), activation='relu', alpha=0.001, 
                           learning_rate='adaptive', max_iter=1000):
        """Train a simple neural network without hyperparameter tuning"""
        print("\n=== Training Simple Neural Network ===")
        
        nn = MLPRegressor(
            hidden_layer_sizes=hidden_layer_sizes,
            activation=activation,
            alpha=alpha,
            learning_rate=learning_rate,
            max_iter=max_iter,
            random_state=42
        )
        
        self.model = make_pipeline(
            SimpleImputer(strategy='mean'),
            StandardScaler(),
            nn
        )
        
        self.model.fit(self.X_train, self.y_train)
        self.best_model = self.model
        self.best_model_name = 'Neural Network'
        
        test_pred = self.model.predict(self.X_test)
        self.best_r2 = r2_score(self.y_test, test_pred)
        
        self.save_best_model()
        return self.model
    
    def cross_validate(self, cv=5):
        """Perform cross-validation"""
        if self.best_model is None:
            raise ValueError("No model trained. Please train a model first.")
            
        print("\n=== Cross-Validation ===")
        r2_scores = cross_val_score(self.best_model, self.X_train, self.y_train, cv=cv, scoring='r2')
        print(f"R² scores: {r2_scores}")
        print(f"Mean R²: {np.mean(r2_scores):.4f} ± {np.std(r2_scores):.4f}")
    
    def save_best_model(self):
        """Save the best performing model to disk"""
        if self.best_model is None:
            raise ValueError("No best model identified yet")
            
        filename = f"best_model_{self.best_model_name.replace(' ', '_')}.pkl"
        filepath = os.path.join(self.output_dir, filename)
        
        model_info = {
            'model': self.best_model,
            'scaler': self.scaler,
            'imputer': self.imputer,
            'feature_names': self.feature_names,
            'model_name': self.best_model_name,
            'test_r2': self.best_r2,
            'best_params': self.best_params
        }
        
        joblib.dump(model_info, filepath)
        print(f"\nSaved best model ({self.best_model_name}) to {filepath}")
    
    def plot_parity(self, model=None):
        """Create and save parity plot for best model"""
        if model is None:
            if self.best_model is None:
                raise ValueError("No best model available")
            model = self.best_model
            model_name = self.best_model_name
        else:
            model_name = "Selected Model"
        
        train_pred = model.predict(self.X_train)
        test_pred = model.predict(self.X_test)
        
        plt.figure(figsize=(8, 6))
        plt.scatter(self.y_train, train_pred, alpha=1, label='Train')
        plt.scatter(self.y_test, test_pred, alpha=1, label='Test')
        
        min_val = min(min(self.y_train), min(self.y_test))
        max_val = max(max(self.y_train), max(self.y_test))
        plt.plot([min_val, max_val], [min_val, max_val], 'k--')
        
        plt.xlabel('Experimental Value')
        plt.ylabel('Predicted Value')
        plt.title(f'Parity Plot - {model_name} (with Temperature)')
        plt.legend()
        plt.grid(True)
        
        plt.savefig(os.path.join(self.output_dir, 'parity_plot.png'), dpi=300, bbox_inches='tight')
        plt.show()
    
    def plot_loss_curve(self):
        """Plot training loss curve for neural network"""
        if self.best_model is None:
            print("No model available")
            return
            
        nn = self.best_model.named_steps['mlpregressor']
        if hasattr(nn, 'loss_curve_'):
            plt.figure(figsize=(8, 5))
            plt.plot(nn.loss_curve_)
            plt.title('Training Loss Curve - Neural Network')
            plt.xlabel('Epochs')
            plt.ylabel('Loss')
            plt.grid(True)
            
            plt.savefig(os.path.join(self.output_dir, 'loss_curve.png'), dpi=300, bbox_inches='tight')
            plt.show()
        else:
            print("Loss curve not available for this model")
    
    def plot_feature_importance(self, top_n=20):
        """Plot feature importance based on neural network weights with adjusted margins to prevent text clipping"""
        if self.best_model is None:
            raise ValueError("No best model available")
        
        nn = self.best_model.named_steps['mlpregressor']
        
        if hasattr(nn, 'coefs_'):
            importance = np.abs(nn.coefs_[0]).sum(axis=1)
            
            idx = np.argsort(importance)[::-1]
            sorted_names = [self.feature_names[i] for i in idx]
            sorted_importance = importance[idx]
            
            plt.figure(figsize=(10, 6))
            plt.barh(range(min(top_n, len(sorted_importance))), 
                     sorted_importance[:top_n][::-1], 
                     align='center')
            plt.yticks(range(min(top_n, len(sorted_importance))), 
                       sorted_names[:top_n][::-1])
            plt.xlabel('Feature Importance (Sum of Absolute Weights)')
            plt.title(f'Top {top_n} Features - Neural Network')
            
            # Add extra space on the left side to avoid cutting off long feature names (like Inverse_Temperature_K)
            plt.subplots_adjust(left=0.3)
            plt.tight_layout()
            
            plt.savefig(os.path.join(self.output_dir, 'feature_importance.png'), dpi=300, bbox_inches='tight')
            plt.show()
        else:
            print("Feature weights not available for this model")

# Example usage
if __name__ == "__main__":
    predictor = Chemical_Property_Predictor_NN(output_dir='best_models')
    
    # Load data using your file name format
    predictor.load_data('Original_data.xlsx')
    
    # Generate descriptors and include temperature
    predictor.generate_descriptors()
    
    # Preprocess data
    predictor.preprocess_data()
    
    # Run hyperparameter optimization or training
    predictor.hyperparameter_tuning(search_type='randomized', cv=5, n_iter=10)
    
    
    # Cross-validation
    predictor.cross_validate()
    
    # Generate visualizations & save outputs automatically
    predictor.plot_parity()
    predictor.plot_loss_curve()
    predictor.plot_feature_importance()
```
