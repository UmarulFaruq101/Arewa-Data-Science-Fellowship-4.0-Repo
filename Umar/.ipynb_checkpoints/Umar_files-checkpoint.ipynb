```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from rdkit import Chem
from rdkit.Chem import AllChem, Descriptors, rdMolDescriptors
from rdkit.Chem.MolStandardize import rdMolStandardize
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
from sklearn.feature_selection import VarianceThreshold
from sklearn.impute import SimpleImputer
from sklearn.linear_model import (
    LinearRegression, Ridge, Lasso, ElasticNet, BayesianRidge,
    SGDRegressor, HuberRegressor, RANSACRegressor, TweedieRegressor
)
from sklearn.svm import SVR, LinearSVR
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import (
    RandomForestRegressor, GradientBoostingRegressor, AdaBoostRegressor,
    BaggingRegressor, ExtraTreesRegressor
)
from sklearn.neighbors import KNeighborsRegressor
from sklearn.neural_network import MLPRegressor
import joblib
import os
import warnings
warnings.filterwarnings('ignore')


class Chemical_Property_Predictor:
    def __init__(self, output_dir='best_models'):
        self.raw_data = None
        self.data = None
        self.temp_data = None
        self.descriptors = None
        self.X = None
        self.y = None
        self.X_train = None
        self.X_test = None
        self.y_train = None
        self.y_test = None
        self.scaler = None
        self.imputer = None
        self.model = None
        self.best_model = None
        self.best_model_name = None
        self.best_r2 = -np.inf
        self.output_dir = output_dir
        self.feature_names = None
        self.X_ = None
        
        os.makedirs(self.output_dir, exist_ok=True)
        
        # Stable models (no problematic ones)
        self.models = {
            'Linear Regression': LinearRegression(),
            'Ridge Regression': Ridge(),
            'Lasso Regression': Lasso(),
            'ElasticNet': ElasticNet(),
            'Bayesian Ridge': BayesianRidge(),
            'SGD Regressor': SGDRegressor(max_iter=1000, tol=1e-3, random_state=42),
            'Huber Regressor': HuberRegressor(),
            'RANSAC': RANSACRegressor(random_state=42),
            'Tweedie': TweedieRegressor(power=0, max_iter=1000),
            'SVR': SVR(),
            'Linear SVR': LinearSVR(max_iter=10000, random_state=42),
            'Decision Tree': DecisionTreeRegressor(random_state=42),
            'Random Forest': RandomForestRegressor(n_estimators=100, random_state=42),
            'Gradient Boosting': GradientBoostingRegressor(n_estimators=100, random_state=42),
            'AdaBoost': AdaBoostRegressor(n_estimators=100, random_state=42),
            'Bagging': BaggingRegressor(n_estimators=100, random_state=42),
            'Extra Trees': ExtraTreesRegressor(n_estimators=100, random_state=42),
            'K-Neighbors': KNeighborsRegressor(),
            'Neural Network': MLPRegressor(hidden_layer_sizes=(10, 5), max_iter=1000, random_state=42)
        }

    def _process_molecule(self, smiles):
        try:
            mol = Chem.MolFromSmiles(smiles)
            if mol is None:
                return None
            mol = Chem.MolFromSmiles(Chem.MolToSmiles(mol))
            if mol is None:
                return None
            parent = rdMolStandardize.FragmentParent(mol)
            mol = Chem.AddHs(parent)
            try:
                AllChem.EmbedMolecule(mol, randomSeed=42)
                AllChem.MMFFOptimizeMolecule(mol)
            except:
                pass
            return mol
        except:
            return None

    def load_data(self, file_path):
        """Load CSV or Excel file and auto‑rename columns to expected names."""
        if file_path.endswith(('.xlsx', '.xls')):
            self.raw_data = pd.read_excel(file_path)
        else:
            self.raw_data = pd.read_csv(file_path)
        
        # Map actual column names to expected ones
        rename_map = {}
        if 'NAME' in self.raw_data.columns:
            rename_map['NAME'] = 'substance_name'
        if 'SMILES' in self.raw_data.columns:
            rename_map['SMILES'] = 'smiles'
        if 'Log_Psat_mmHg' in self.raw_data.columns:
            rename_map['Log_Psat_mmHg'] = 'property'
        # Temperature columns already have correct names, no rename needed
        
        if rename_map:
            self.raw_data = self.raw_data.rename(columns=rename_map)
        
        # Now check all required columns exist
        required = ['substance_name', 'smiles', 'property', 'Inverse_Temperature_K', 'Temperature_C']
        for col in required:
            if col not in self.raw_data.columns:
                raise ValueError(f"Column '{col}' not found. Found: {list(self.raw_data.columns)}")
        
        self.data = self.raw_data[['substance_name', 'smiles', 'property']].copy()
        self.temp_data = self.raw_data[['Inverse_Temperature_K', 'Temperature_C']].copy()
        print(f"Loaded {len(self.data)} rows from {file_path}")

    def generate_descriptors(self):
        if self.data is None:
            raise ValueError("Load data first.")
        descriptor_list = []
        temp_list = []
        valid_indices = []
        for idx, row in self.data.iterrows():
            mol = self._process_molecule(row['smiles'])
            if mol is not None:
                desc = {}
                for name, func in Descriptors.descList:
                    try:
                        desc[name] = func(mol)
                    except:
                        desc[name] = np.nan
                desc['NumAtoms'] = mol.GetNumAtoms()
                desc['NumHeavyAtoms'] = mol.GetNumHeavyAtoms()
                desc['NumRotatableBonds'] = rdMolDescriptors.CalcNumRotatableBonds(mol)
                desc['NumHDonors'] = rdMolDescriptors.CalcNumHBD(mol)
                desc['NumHAcceptors'] = rdMolDescriptors.CalcNumHBA(mol)
                desc['TPSA'] = rdMolDescriptors.CalcTPSA(mol)
                desc['MolLogP'] = Descriptors.MolLogP(mol)
                desc['MolWt'] = Descriptors.MolWt(mol)
                desc['RingCount'] = rdMolDescriptors.CalcNumRings(mol)
                try:
                    desc['PMI1'] = rdMolDescriptors.CalcPMI1(mol)
                    desc['PMI2'] = rdMolDescriptors.CalcPMI2(mol)
                    desc['PMI3'] = rdMolDescriptors.CalcPMI3(mol)
                    desc['RadiusOfGyration'] = rdMolDescriptors.CalcRadiusOfGyration(mol)
                except:
                    pass
                descriptor_list.append(desc)
                temp_list.append(self.temp_data.loc[idx].values)
                valid_indices.append(idx)
            else:
                print(f"Could not process SMILES: {row['smiles']}")
        self.descriptors = pd.DataFrame(descriptor_list, index=valid_indices)
        self.temp_data = pd.DataFrame(temp_list, 
                                      columns=['Inverse_Temperature_K', 'Temperature_C'],
                                      index=valid_indices)
        self.data = self.data.loc[valid_indices]
        print(f"Generated {self.descriptors.shape[1]} molecular descriptors + 2 temperature features.")

    def preprocess_data(self, test_size=0.2, random_state=42, strategy='mean'):
        if self.descriptors is None:
            raise ValueError("Generate descriptors first.")
        X_combined = pd.concat([self.descriptors, self.temp_data], axis=1)
        selector = VarianceThreshold(threshold=0.001)
        selected = selector.fit_transform(X_combined)
        self.feature_names = np.array(X_combined.columns)[selector.get_support()]
        X_filtered = pd.DataFrame(selected, columns=self.feature_names)
        for temp_col in ['Inverse_Temperature_K', 'Temperature_C']:
            if temp_col not in self.feature_names:
                X_filtered[temp_col] = X_combined[temp_col].values
                self.feature_names = list(self.feature_names) + [temp_col]
        self.X = X_filtered
        self.y = self.data['property']
        self.X_train, self.X_test, self.y_train, self.y_test = train_test_split(
            self.X, self.y, test_size=test_size, random_state=random_state)
        self.imputer = SimpleImputer(strategy=strategy)
        self.scaler = StandardScaler()
        self.X_train = pd.DataFrame(
            self.scaler.fit_transform(self.imputer.fit_transform(self.X_train)),
            columns=self.feature_names
        )
        self.X_test = pd.DataFrame(
            self.scaler.transform(self.imputer.transform(self.X_test)),
            columns=self.feature_names
        )
        self.X_ = self.X
        print(f"Train: {len(self.y_train)}, Test: {len(self.y_test)}")

    def evaluate_all_models(self):
        print("\n=== Evaluating All Models (with Temperature) ===")
        results = []
        for name, model in self.models.items():
            try:
                print(f"\nTraining {name}...")
                model.fit(self.X_train, self.y_train)
                pred = model.predict(self.X_test)
                r2 = r2_score(self.y_test, pred)
                rmse = np.sqrt(mean_squared_error(self.y_test, pred))
                results.append((name, r2, rmse))
                if r2 > self.best_r2:
                    self.best_r2 = r2
                    self.best_model = model
                    self.best_model_name = name
                    print(f"  >>> New best: {name} (R²={r2:.3f})")
            except Exception as e:
                print(f"  {name} failed: {str(e)[:80]}...")
        results.sort(key=lambda x: x[1], reverse=True)
        print("\n=== Final Rankings ===")
        print("{:<25} {:<10} {:<10}".format("Model", "R²", "RMSE"))
        for name, r2, rmse in results:
            print("{:<25} {:<10.3f} {:<10.3f}".format(name, r2, rmse))
        self.save_best_model()
        return results

    def save_best_model(self):
        if self.best_model is None:
            print("No best model found.")
            return
        filename = f"best_model_{self.best_model_name.replace(' ', '_')}.pkl"
        filepath = os.path.join(self.output_dir, filename)
        joblib.dump({
            'model': self.best_model,
            'scaler': self.scaler,
            'imputer': self.imputer,
            'feature_names': self.feature_names,
            'model_name': self.best_model_name,
            'test_r2': self.best_r2
        }, filepath)
        print(f"\nBest model saved to: {filepath}")

    def plot_parity(self):
        if self.best_model is None:
            print("Train a model first.")
            return
        train_pred = self.best_model.predict(self.X_train)
        test_pred = self.best_model.predict(self.X_test)
        plt.figure(figsize=(8,6))
        plt.scatter(self.y_train, train_pred, alpha=0.5, label='Train')
        plt.scatter(self.y_test, test_pred, alpha=0.5, label='Test')
        lims = [min(self.y.min(), train_pred.min(), test_pred.min()),
                max(self.y.max(), train_pred.max(), test_pred.max())]
        plt.plot(lims, lims, 'k--')
        plt.xlabel('True Property')
        plt.ylabel('Predicted Property')
        plt.title(f'Parity Plot – {self.best_model_name} (with Temperature)')
        plt.legend()
        plt.grid()
        plt.show()

    def plot_feature_importance(self, top_n=20):
        if self.best_model is None:
            print("No model available.")
            return
        if hasattr(self.best_model, 'coef_'):
            imp = np.abs(self.best_model.coef_)
        elif hasattr(self.best_model, 'feature_importances_'):
            imp = self.best_model.feature_importances_
        else:
            print("No feature importance available.")
            return
        idx = np.argsort(imp)[::-1][:top_n]
        plt.figure(figsize=(10,6))
        plt.barh(range(len(idx)), imp[idx][::-1])
        plt.yticks(range(len(idx)), [self.feature_names[i] for i in idx][::-1])
        plt.xlabel('Importance')
        plt.title(f'Top {top_n} Features – {self.best_model_name}')
        plt.tight_layout()
        plt.show()


if __name__ == "__main__":
    predictor = Chemical_Property_Predictor()
    predictor.load_data('Original_data.xlsx')   # <-- use your file name
    predictor.generate_descriptors()
    predictor.preprocess_data()
    predictor.evaluate_all_models()
    predictor.plot_parity()
    predictor.plot_feature_importance()
```

```python
predictor = Chemical_Property_Predictor()
predictor.load_data('Original_data.xlsx')   # or .csv
predictor.generate_descriptors()
predictor.preprocess_data()
predictor.evaluate_all_models()
predictor.plot_parity()
predictor.plot_feature_importance()
```

```python
from sklearn.inspection import permutation_importance

# Use the best model already stored
result = permutation_importance(
    predictor.best_model,
    predictor.X_test,
    predictor.y_test,
    n_repeats=10,
    random_state=42
)

# Plot top 20
importances = result.importances_mean
idx = np.argsort(importances)[::-1][:20]
plt.figure(figsize=(10,6))
plt.barh(range(len(idx)), importances[idx][::-1])
plt.yticks(range(len(idx)), [predictor.feature_names[i] for i in idx][::-1])
plt.xlabel('Permutation Importance')
plt.title(f'Feature Importance – {predictor.best_model_name}')
plt.tight_layout()
plt.show()
```

```python
from sklearn.inspection import permutation_importance
import seaborn as sns

# Compute importance for the best model (neural network)
result = permutation_importance(
    predictor.best_model,
    predictor.X_test,
    predictor.y_test,
    n_repeats=10,
    random_state=42
)

# Get indices of top 10 features
top_idx = np.argsort(result.importances_mean)[::-1][:10]
top_features = [predictor.feature_names[i] for i in top_idx]

print("Top 10 features:", top_features)
```

```python
# Get the training data (already scaled and imputed)
X_train_df = predictor.X_train

# Select only the top 10 columns
X_top = X_train_df.iloc[:, top_idx]
X_top.columns = top_features
```

```python
plt.figure(figsize=(10, 8))
corr = X_top.corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', center=0, fmt='.2f')
plt.title('Correlation Heatmap – Top 10 Most Important Features')
plt.tight_layout()
plt.show()
```
