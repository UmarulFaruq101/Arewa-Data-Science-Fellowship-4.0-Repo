```python
import pandas as pd

# Load the dataset
df = pd.read_excel('Original_data.xlsx')

# Drop rows where Log_Psat_mmHg equals 1
filtered_df = df[df['Log_Psat_mmHg'] != 1]

# Save the filtered dataset to a new Excel file
filtered_df.to_excel('Original_data_filtered.xlsx', index=False)

print(f"Dropped {len(df) - len(filtered_df)} rows where Log_Psat_mmHg was 1.")
```

```python
import pandas as pd

# Load the dataset
df = pd.read_excel('Original_data.xlsx')

# Drop rows where Log_Psat_mmHg is exactly 1 (or close to 1 using a tolerance if needed, e.g., float precision)
filtered_df = df[abs(df['Log_Psat_mmHg'] - 1) > 1e-5]

# Save the updated file
filtered_df.to_excel('Original_data_filtered.xlsx', index=False)

print(f"Original rows: {len(df)}")
print(f"Rows after dropping: {len(filtered_df)}")
print(f"Successfully dropped {len(df) - len(filtered_df)} rows.")
```

```python
import pandas as pd

# Load the dataset
df = pd.read_excel('Original_data.xlsx')

# Drop rows where Log_Psat_mmHg equals 1 (using a small tolerance for floating-point precision)
filtered_df = df[abs(df['Log_Psat_mmHg'] - 1) > 1e-5]

# Save the filtered dataset to an Excel file
filtered_df.to_excel('Original_data_filtered.xlsx', index=False)

print(f"Dataset successfully saved as 'Original_data_filtered.xlsx'.")
print(f"Total rows remaining: {len(filtered_df)}")
```
