```python
# 1. The following is a list of 10 students ages
ages = [19, 22, 19, 24, 20, 25, 26, 24, 25, 24]

# Sort the list and find the min and max age
ages.sort()
min_age = min(ages)
max_age = max(ages)
print('Sorted:', ages)
print('Min age:', min_age, '| Max age:', max_age)
```

```python
# Add the min age and the max age again to the list
ages.append(min_age)
ages.append(max_age)
print(ages)
```

```python
# Find the median age (one middle item or two middle items divided by two)
sorted_ages = sorted(ages)
n = len(sorted_ages)
if n % 2 == 1:
    median_age = sorted_ages[n // 2]
else:
    median_age = (sorted_ages[n // 2 - 1] + sorted_ages[n // 2]) / 2
print('Median age:', median_age)
```

```python
# Find the average age (sum of all items divided by their number)
average_age = sum(ages) / len(ages)
print('Average age:', average_age)
```

```python
# Find the range of the ages (max minus min)
age_range = max(ages) - min(ages)
print('Range:', age_range)
```

```python
# Compare the value of (min - average) and (max - average), use abs() method
min_diff = abs(min(ages) - average_age)
max_diff = abs(max(ages) - average_age)
print('abs(min - average):', min_diff)
print('abs(max - average):', max_diff)
print('Are they equal?', min_diff == max_diff)
```

```python
# 2. Find the middle country(ies) in the countries list
countries = ['Nigeria', 'Ghana', 'Kenya', 'Egypt', 'Morocco', 'Senegal', 'Tanzania', 'Ethiopia']
n = len(countries)
if n % 2 == 0:
    middle_countries = countries[n // 2 - 1: n // 2 + 1]
else:
    middle_countries = [countries[n // 2]]
print(middle_countries)
```

```python
# 3. Divide the countries list into two equal lists if it is even; if not, one more country for the first half
half = (n + 1) // 2
first_half = countries[:half]
second_half = countries[half:]
print(first_half)
print(second_half)
```
